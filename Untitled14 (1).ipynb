```python
# ==========================================
# CELL 1: Setup and Environment Check
# ==========================================
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNet
from tensorflow.keras.preprocessing import image_dataset_from_directory
import matplotlib.pyplot as plt
import numpy as np
import os

print("Using TensorFlow version:", tf.__version__)
print("GPU Available:", "Yes" if tf.config.list_physical_devices('GPU') else "No")
```

```python
# ==========================================
# CELL 2: Download and Extract Dataset (TrashNet)
# ==========================================
# Downloading the standard TrashNet dataset via GitHub / Kaggle mirror
!git clone https://github.com/garythung/trashnet.git
dataset_dir = 'trashnet/data/dataset-resized'

if not os.path.exists(dataset_dir):
    # Fallback structure check
    print("Please verify dataset path. Looking in:", dataset_dir)
else:
    print("Dataset successfully located at:", dataset_dir)
```

```python
# ==========================================
# CELL 3: Data Loading & Splitting (80/20 Train/Test)
# ==========================================
import urllib.request
import zipfile
import os

IMG_SIZE = 224
BATCH_SIZE = 32
dataset_dir = 'trashnet/data/dataset-resized'

# Automatically download and extract the dataset if it's missing
if not os.path.exists(dataset_dir):
    print("Dataset directory not found. Downloading TrashNet dataset-resized.zip...")
    os.makedirs('trashnet/data', exist_ok=True)
    zip_url = "https://raw.githubusercontent.com/garythung/trashnet/master/data/dataset-resized.zip"
    zip_path = "trashnet/data/dataset-resized.zip"

    try:
        urllib.request.urlretrieve(zip_url, zip_path)
        print("Download complete! Extracting dataset...")
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall('trashnet/data')
        print("Extraction complete!")
    except Exception as e:
        # Fallback to a secondary mirror if the main raw link fails
        print("Primary mirror failed. Attempting alternative download source...")
        alt_url = "https://github.com/garythung/trashnet/raw/master/data/dataset-resized.zip"
        urllib.request.urlretrieve(alt_url, zip_path)
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall('trashnet/data')
        print("Alternative extraction complete!")

train_dataset = image_dataset_from_directory(
    dataset_dir,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE
)

validation_dataset = image_dataset_from_directory(
    dataset_dir,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE
)

class_names = train_dataset.class_names
print("Detected Waste Classes:", class_names)

# Optimize performance with prefetching
AUTOTUNE = tf.data.AUTOTUNE
train_dataset = train_dataset.prefetch(buffer_size=AUTOTUNE)
validation_dataset = validation_dataset.prefetch(buffer_size=AUTOTUNE)
```

```python
# ==========================================
# CELL 4: Build Model via Transfer Learning (MobileNet)
# ==========================================
base_model = MobileNet(input_shape=(IMG_SIZE, IMG_SIZE, 3),
                       include_top=False,
                       weights='imagenet')

# Freeze the pretrained base model
base_model.trainable = False

# Add custom classification head matching our trash categories
global_average_layer = layers.GlobalAveragePooling2D()
prediction_layer = layers.Dense(len(class_names), activation='softmax')

inputs = tf.keras.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
x = tf.keras.applications.mobilenet.preprocess_input(inputs)
x = base_model(x, training=False)
x = global_average_layer(x)
x = layers.Dropout(0.2)(x)
outputs = prediction_layer(x)

model = tf.keras.Model(inputs, outputs)

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.summary()
```

```python
# ==========================================
# CELL 5: Train the Model
# ==========================================
EPOCHS = 5

history = model.fit(
    train_dataset,
    validation_data=validation_dataset,
    epochs=EPOCHS
)
```

```python
# ==========================================
# CELL 6: Evaluate Metrics & Generate Results
# ==========================================
loss, accuracy = model.evaluate(validation_dataset)
print(f"\nValidation Accuracy achieved: {accuracy * 100:.2f}%")
print(f"Validation Loss: {loss:.4f}")

# Ensure the target folder exists
os.makedirs('results', exist_ok=True)

# Plot training history curves for your 'results/' folder
plt.figure(figsize=(8, 5))
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Training vs Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.savefig('results/accuracy_curve.png')
plt.show()
```
